```python
musteriler = pd.DataFrame({
    "musteri_id": [1, 2, 3, 4, 5],
    "isim": ["Ali", "Ayşe", "Emre", "Zeynep", "Fulya"]
})

siparisler = pd.DataFrame({
    "siparis_id": [101, 102, 103, 104, 105],
    "musteri_id": [1, 2, 1, 6, 7], # 6 ve 7 ID'li müşteriler musteriler tablosunda yok
    "urun": ["Laptop", "Telefon", "Tablet", "Klavye", "Fare"],
    "tutar": [2500, 1200, 800, 300, 150]
})

# 9.1 Inner Join (Sadece ortak olanlar: 1 ve 2 ID'li 2 müşteri eşleşir)
merge_inner = pd.merge(musteriler, siparisler, on="musteri_id", how="inner")
print("9.1 Inner Join (Eşleşen müşteri sayısı: {}):\n".format(len(merge_inner)), merge_inner)

# 9.2 Left Join (Siparişi olmayan Emre ve Fulya için urun ve tutar NaN gelir)
merge_left = pd.merge(musteriler, siparisler, on="musteri_id", how="left")
print("\n9.2 Left Join:\n", merge_left)

# 9.3 Outer Join (Her iki tablodaki tüm kayıtlar gelir; eşleşmeyenler NaN olur)
merge_outer = pd.merge(musteriler, siparisler, on="musteri_id", how="outer")
print("\n9.3 Outer Join:\n", merge_outer)
```

```python
hafta1 = pd.DataFrame({"gun": ["Pzt", "Sal"], "adet": [20, 35]})
hafta2 = pd.DataFrame({"gun": ["Çar", "Per"], "adet": [28, 40]})
gelir = pd.DataFrame({"toplam_gelir": [1000, 1750]})

# 10.1 Alt alta birleştirme
birlestir_dikey = pd.concat([hafta1, hafta2], ignore_index=True)
print("10.1 Alt alta (Satır bazlı) Concat:\n", birlestir_dikey)

# 10.2 Yan yana birleştirme
birlestir_yatay = pd.concat([hafta1, gelir], axis=1)
print("\n10.2 Yan yana (Sütun bazlı) Concat:\n", birlestir_yatay)
```

```python
df_calisan = pd.DataFrame({
    "sube": ["A", "B", "A", "C", "B", "A", "C"],
    "calisan": ["Ayşe", "Burak", "Cem", "Deniz", "Elif", "Fatih", "Gizem"],
    "satis": [1500, 2000, 1200, 1800, 2200, 1000, 2500]
})
print("Çalışan Tablosu:\n", df_calisan)

# 11.1 Toplam Satış
toplam_satis = df_calisan.groupby("sube")["satis"].sum()
print("\n11.1 Şube Bazında Toplam Satış:\n", toplam_satis)

# 11.2 Ortalama Satış
ortalama_satis = df_calisan.groupby("sube")["satis"].mean()
print("\n11.2 Şube Bazında Ortalama Satış:\n", ortalama_satis)

# 11.3 Birden fazla istatistik (.agg)
ozet_istatistik = df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"])
print("\n11.3 Şube Başına Çoklu İstatistik:\n", ozet_istatistik)
```

```python
df_satis = pd.DataFrame({
    "sehir": ["İstanbul", "Ankara", "İzmir", "İstanbul", "Ankara", "İzmir", "İstanbul"],
    "kategori": ["Elektronik", "Giyim", "Ev Aletleri", "Elektronik", "Giyim", "Ev Aletleri", "Giyim"],
    "satis": [1500, 800, 500, 2000, 1000, 750, 900]
})
print("Orijinal Satış Tablosu:\n", df_satis)

# 12.1 Temel Pivot Tablo
pivot_1 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis")
print("\n12.1 Temel Pivot:\n", pivot_1)

# 12.2 aggfunc='sum' ve fill_value=0 ile eksik verileri sıfırlama
pivot_2 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis", aggfunc="sum", fill_value=0)
print("\n12.2 Doldurulmuş Pivot:\n", pivot_2)
```

```python
df_gelir = pd.DataFrame({
    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],
    "gelir": [10000, 12000, 11000, 13000],
    "gider": [5000, 6000, 5500, 6500]
})
print("Gelir ve Gider Tablosu:\n", df_gelir)

# 15.1 Çizgi Grafiği
plt.figure(figsize=(6, 4))
df_gelir.plot(x="ceyrek", y="gelir", kind="line", marker="o", title="Çeyreklik Gelir Grafiği")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.grid(True)
plt.show()

# 15.2 Bar Grafiği
plt.figure(figsize=(6, 4))
df_gelir.plot(x="ceyrek", y="gelir", kind="bar", color="skyblue", title="Çeyreklik Gelir Dağılımı")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.xticks(rotation=0)
plt.show()

# 15.3 Pasta Grafiği
plt.figure(figsize=(5, 5))
plt.pie(df_gelir["gelir"], labels=df_gelir["ceyrek"], autopct="%1.1f%%", startangle=90)
plt.title("Çeyreklik Gelir Payı")
plt.show()

# 15.4 Scatter Grafiği ve Yorum
plt.figure(figsize=(6, 4))
df_gelir.plot(x="gelir", y="gider", kind="scatter", color="red", title="Gelir - Gider İlişkisi")
plt.xlabel("Gelir (TL)")
plt.ylabel("Gider (TL)")
plt.grid(True)
plt.show()

# Yorum:
print("Grafik Yorumu: Gelir arttıkça giderlerin de düzenli olarak arttığı doğrusal ve pozitif yönlü bir ilişki gözlemlenmektedir.")
```
