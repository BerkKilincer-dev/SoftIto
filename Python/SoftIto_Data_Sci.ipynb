```python
import pandas as pd

# Versiyon kontrolü
print(pd.__version__)
```

```python
# Basit bir Series oluşturma
seri = pd.Series([100, 200, 300, 400])
print(seri)

# Şehir ve nüfus Series'i
nufus = pd.Series(
    [15840000, 5800000, 4470000, 3215000],
    index=["İstanbul", "Ankara", "İzmir", "Bursa"]
)
print(nufus)

# Index adıyla seçim yapma
print("İzmir Nüfusu:", nufus["İzmir"])

# Toplam nüfusu bulma
print("Toplam Nüfus:", nufus.sum())
```

```python
# Ürün, fiyat, stok tablosu (5 satır)
urunler_df = pd.DataFrame({
    "urun": ["Laptop", "Mouse", "Klavye", "Monitör", "Kulaklık"],
    "fiyat": [28000, 450, 1200, 6500, 1800],
    "stok": [15, 120, 60, 25, 80]
})
print(urunler_df)

# Öğrenci bilgileri tablosu
ogrenciler_df = pd.DataFrame({
    "isim": ["Ali", "Ayşe", "Mehmet", "Zeynep"],
    "yas": [21, 22, 20, 23],
    "bolum": ["Yazılım Müh.", "Bilgisayar Müh.", "Endüstri Müh.", "Yazılım Müh."]
})
print(ogrenciler_df)

# Satır ve sütun sayısını kontrol etme (çıktı: (4, 3))
print("Satır ve Sütun Boyutu:", ogrenciler_df.shape)
```

```python
# index=False ile temiz bir şekilde kaydetme
ogrenciler_df.to_csv("ogrenciler.csv", index=False)

# Kaydedilen dosyayı okuma
okunan_df = pd.read_csv("ogrenciler.csv")
print(okunan_df)

# index=False parametresi olmadan kaydetme farkı:
ogrenciler_df.to_csv("ogrenciler_indexli.csv")
# Not: index=False yazılmazsa Pandas varsayılan satır indekslerini (0, 1, 2...)
# CSV'ye ayrı bir sütun ("Unnamed: 0") olarak yazar.
```

```python
ornek_df = pd.DataFrame({
    "isim": ["Can", "Ece", "Burak", "Selin", "Deniz", "Kaan"],
    "yas": [24, 30, 19, 42, 28, 35],
    "maas": [22000, 31000, 18000, 45000, 29000, 34000]
})

print(ornek_df.head(3))     # İlk 3 satır
print(ornek_df.tail(2))     # Son 2 satır
print("Boyut:", ornek_df.shape)  # (satır, sütun)

ornek_df.info()             # Veri tipleri ve boş değer kontrolü
print(ornek_df["maas"].describe())  # Ortalama, min, max, çeyreklikler
```

```python
# Tek sütun seçimi (Series döner)
print(ornek_df["isim"])

# Birden fazla sütun seçimi (DataFrame döner)
print(ornek_df[["isim", "yas"]])

# .loc ile ilk 3 satır (etiket bazlı: 0, 1 ve 2 dahil)
print(ornek_df.loc[0:2])

# .iloc ile 2. satır (indeks 1) ve 2. sütun (indeks 1) kesişimi
print(ornek_df.iloc[1, 1])
```

```python
filtre_df = pd.DataFrame({
    "isim": ["Ahmet", "Elif", "Murat", "Buse", "Oğuz"],
    "yas": [26, 19, 32, 45, 23],
    "maas": [25000, 18000, 29000, 42000, 21000],
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Bursa"]
})

# Yaşı 25'ten büyük olanlar
print(filtre_df[filtre_df["yas"] > 25])

# Maaşı 20.000 ile 30.000 arasında olanlar
print(filtre_df[(filtre_df["maas"] >= 20000) & (filtre_df["maas"] <= 30000)])

# Şehri "İstanbul" olanlar
print(filtre_df[filtre_df["sehir"] == "İstanbul"])

# Yaşı 20'den küçük VEYA 40'tan büyük olanlar
print(filtre_df[(filtre_df["yas"] < 20) | (filtre_df["yas"] > 40)])
```

```python
# Doğum yılı sütunu ekleme
filtre_df["dogum_yili"] = 2026 - filtre_df["yas"]

# Maaşları %10 artırma
filtre_df["maas"] = filtre_df["maas"] * 1.10

# apply ve lambda ile yaş durumu sütunu
filtre_df["durum"] = filtre_df["yas"].apply(lambda x: "Yetişkin" if x >= 18 else "Çocuk")

print(filtre_df)
```

```python
# Yaşa göre küçükten büyüğe
print(filtre_df.sort_values(by="yas"))

# Maaşa göre büyükten küçüğe
print(filtre_df.sort_values(by="maas", ascending=False))

# İki sütuna göre sıralama (önce sehir A-Z, sonra yas küçükten büyüğe)
print(filtre_df.sort_values(by=["sehir", "yas"], ascending=[True, True]))
```

```python
sinif_df = pd.DataFrame({
    "isim": ["Bora", "Ceren", "Demir", "Ezgi", "Fırat", "Gamze"],
    "yas": [22, 20, 24, 21, 23, 19],
    "sehir": ["Ankara", "İstanbul", "İzmir", "Bursa", "Eskişehir", "Antalya"],
    "not1": [85, 60, 90, 65, 75, 55],
    "not2": [78, 65, 88, 70, 80, 60],
    "not3": [92, 70, 85, 72, 70, 58]
})

# Üç dersin ortalamasını hesaplayıp yeni sütun yapma
sinif_df["ortalama"] = (sinif_df["not1"] + sinif_df["not2"] + sinif_df["not3"]) / 3

# Ortalaması 70'in üzerinde olanları filtreleyip yaşa göre küçükten büyüğe sıralama
sonuc = sinif_df[sinif_df["ortalama"] > 70].sort_values(by="yas", ascending=True)

print(sonuc)
```
