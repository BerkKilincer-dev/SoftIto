```python
# 19
def sayi_durumu(sayi):
    if sayi > 0:
        return "Pozitif"
    elif sayi < 0:
        return "Negatif"
    else:
        return "Sıfır"

print("-8'in durumu:", sayi_durumu(-8))

def en_buyuk_bul(a, b, c):
    en_buyuk = a
    if b > en_buyuk:
        en_buyuk = b
    if c > en_buyuk:
        en_buyuk = c
    return en_buyuk

print("12, 45, 23 arasından en büyüğü:", en_buyuk_bul(12, 45, 23))
```

```python
#22
notlar = []
toplam = 0

for i in range(4):
    ders_notu = float(input(f"{i+1}. dersin notunu girin: "))
    notlar.append(ders_notu)
    toplam += ders_notu

ortalama = toplam / 4
print(f"\nNot Ortalamanız: {ortalama:.2f}")

if ortalama >= 90:
    harf = "AA"
elif ortalama >= 80:
    harf = "BA"
elif ortalama >= 70:
    harf = "BB"
elif ortalama >= 60:
    harf = "CB"
elif ortalama >= 50:
    harf = "CC"
else:
    harf = "FF"

print(f"Harf Notunuz: {harf}")


print(f"En yüksek notunuz: {max(notlar)}")
print(f"En düşük notunuz: {min(notlar)}")
```

```python
#23
sepet_fiyatlari = [45.50, 120.00, 30.00, 250.00, 15.75, 110.00]

print("100 TL Üzeri Ürünler:")
for fiyat in sepet_fiyatlari:
    if fiyat > 100:
        print(f"- {fiyat} TL")


toplam_tutar = sum(sepet_fiyatlari)
print(f"\nSepet Toplamı: {toplam_tutar} TL")

if toplam_tutar > 500:
    print("Tebrikler, 500 TL üzerindeki alışverişiniz için %10 indirim kazandınız!")
    indirimli_tutar = toplam_tutar * 0.90
    print(f"İndirimli Ödenecek Tutar: {indirimli_tutar:.2f} TL")
else:
    eksik_tutar = 500 - toplam_tutar
    print(f"%10 indirim kazanmak için {eksik_tutar:.2f} TL tutarında daha alışveriş yapmalısınız.")
```
